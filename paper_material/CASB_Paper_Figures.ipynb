```python
from pathlib import Path
import sys, json, hashlib
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, Circle
from matplotlib.lines import Line2D
from scipy.stats import wilcoxon, rankdata, friedmanchisquare, binomtest
from sklearn.neighbors import NearestNeighbors
from threadpoolctl import threadpool_limits
from IPython.display import display, Image
ROOT=Path.cwd()
if not (ROOT/'binary_oversampling.py').exists(): ROOT=ROOT.parent
sys.path.insert(0,str(ROOT))
from binary_oversampling import sampler, CASBSMOTE
thread_guard=threadpool_limits(limits=1)
OUT=ROOT/'paper_material';OUT.mkdir(exist_ok=True)
plt.rcParams.update({'font.family':'DejaVu Sans','font.size':11,'axes.titlesize':13,
 'pdf.fonttype':42,'ps.fonttype':42,'svg.fonttype':'none','savefig.facecolor':'white'})
NAVY='#17324D';BLUE='#2166AC';ORANGE='#D55E00';GREEN='#008B70';PURPLE='#8055A0';GRAY='#9BA6B0';LIGHT='#EFF4F7'
captions=[];figures=[]
def save(fig,stem,title,caption):
    for ext in ['png','pdf','svg']:
        fig.savefig(OUT/(stem+'.'+ext),dpi=300,bbox_inches='tight',facecolor='white',metadata={'Creator':'Matplotlib'} if ext=='pdf' else None)
    plt.close(fig);figures.append(stem);captions.append((title,caption))
    display(Image(filename=str(OUT/(stem+'.png')),width=1000))
def box(ax,x,y,w,h,text,color=BLUE,fs=11):
    ax.add_patch(FancyBboxPatch((x,y),w,h,boxstyle='round,pad=0.012,rounding_size=0.015',fc=LIGHT,ec=color,lw=1.3))
    ax.text(x+w/2,y+h/2,text,ha='center',va='center',fontsize=fs,color=NAVY,linespacing=1.4)
def arrow(ax,start,end,color=GRAY):
    ax.annotate('',xy=end,xytext=start,arrowprops={'arrowstyle':'-|>','color':color,'lw':1.6,'shrinkA':3,'shrinkB':3})
def bare(figsize=(14,8)):
    f,a=plt.subplots(figsize=figsize);a.set(xlim=(0,1),ylim=(0,1));a.axis('off');return f,a

# Figure 1: original schematic, not a reproduction of UGO artwork.
fig,ax=bare((15,9))
ax.text(.02,.97,'Established approaches and the CASB design',fontsize=22,weight='bold',color=NAVY)
rows=[('SMOTE',['Minority source\nstandard selection','Minority neighbor\nline interpolation','Add synthetic\nminority samples']),
 ('Borderline-SMOTE',['Identify minority\ndanger observations','Interpolate from\ndanger sources','Add synthetic\nminority samples']),
 ('UGO',['Pre-detect noise;\nfit classifier; MMA','Select uncertain seeds;\nplug-in generator','Check / relocate noise;\nretrain until stopping']),
 ('CASB family',['Coverage / border / rare\nsource pools + quotas','Capped / unrestricted\ninterpolation or copies','Optional majority ENN\n(CASB-edited only)'])]
for i,(name,stages) in enumerate(rows):
    y=.76-i*.205
    ax.text(.015,y+.065,name,fontsize=13,weight='bold',color=GREEN if i==3 else NAVY,va='center')
    for j,t in enumerate(stages):
        x=.2+j*.27;box(ax,x,y,.23,.135,t,GREEN if i==3 else BLUE,11)
        if j<2:arrow(ax,(x+.23,y+.068),(x+.27,y+.068))
ax.text(.02,.035,'CASB changes allocation and generation control. It does not implement UGO\'s uncertainty loop or noise relocation.',fontsize=11,color=NAVY)
save(fig,'01_prior_work_and_casb','Prior mechanisms and CASB',
 'Original schematic comparing established pipelines with the current CASB family. UGO is summarized from Saglam and Cengiz (2026), Sections 3.1-3.4 and Figs. 2-4; it was not reimplemented or benchmarked here. CASB is independently defined by its neighborhood pools, quotas and generator options. These differences are descriptive, not proof of novelty. SMOTE: Chawla et al. (2002), doi:10.1613/jair.953. Borderline-SMOTE: Han et al. (2005), doi:10.1007/11538059_91. UGO: doi:10.1016/j.knosys.2026.115886.')

fig,ax=bare((15,9))
ax.text(.02,.96,'CASB family: training-time workflow',fontsize=22,weight='bold',color=NAVY)
top=['Training partition\nonly','Fit imputation / scaling\nencode categorical inputs','Compute local geometry\ncoverage, border, rare pools']
for x,w,t in zip([.025,.35,.675],[.27,.27,.30],top):box(ax,x,.77,w,.12,t)
arrow(ax,(.295,.83),(.35,.83));arrow(ax,(.62,.83),(.675,.83))
box(ax,.25,.56,.50,.12,'Target added count: max(0, ceil(ratio × n_majority) - n_minority)\nInteger branch quotas; sample sources with replacement',GREEN)
arrow(ax,(.825,.77),(.65,.68))
variants=[('CASB','50 / 30 / 20\nConditional interpolation caps'),('CASB-relaxed','70 / 20 / 10\nUnrestricted interpolation'),('CASB-edited','70 / 20 / 10\nUnrestricted + majority ENN'),('CASB-ROS','50 / 30 / 20\nDuplicate selected sources')]
for i,(name,t) in enumerate(variants):
    x=.025+.245*i;box(ax,x,.30,.21,.17,name+'\n'+t,GREEN,10)
    arrow(ax,(.5,.56),(x+.105,.47));arrow(ax,(x+.105,.30),(.5,.20))
box(ax,.25,.08,.50,.12,'Fit SVM, KNN, RF or LR on the resampled training set\nEvaluate untouched held-out observations',BLUE)
ax.text(.025,.015,'Ratios are chosen by inner training validation; resampling and PCA never use held-out test observations.',fontsize=10,color=NAVY)
save(fig,'02_casb_family_workflow','CASB family workflow',
 'Workflow for the implemented variants. Percentages are quotas for newly added observations, not disjoint fractions of the original data. Pools overlap; an empty branch draws sources from coverage while retaining its branch-specific interpolation rule. Current experiments use no optional CASB majority pre-drop. The sampling ratio is selected using the existing inner-validation objective; the three ratio candidates are 0.5, 0.75 and 1.0. ENN is majority-only, three neighbors, kind_sel="mode". Preprocessing is refit within each training split.')

# Figure 3: exact geometry of an illustrative capped case.
fig,axes=plt.subplots(1,3,figsize=(15,5),sharex=True,sharey=True)
for ax,(title,limit,color) in zip(axes,[('SMOTE: full selected segment',1.,ORANGE),('CASB: eligible capped branch',.25,GREEN),('CASB-relaxed: full segment',1.,ORANGE)]):
    ax.plot([0,4],[0,0],color=GRAY,lw=2)
    ax.plot([0,4*limit],[0,0],color=color,lw=7,alpha=.5)
    ax.scatter([0,4],[0,0],c=BLUE,s=100,zorder=3)
    ax.scatter([0],[2],c=GRAY,s=90,zorder=3)
    ax.plot([0,0],[0,2],'--',color=GRAY)
    ax.text(.12,1.1,'d = 2',fontsize=11)
    ax.scatter(np.linspace(.1,4*limit-.1,6),np.zeros(6),c=color,marker='^',s=48,zorder=4)
    ax.text(0,-.42,'Source $x_i$',ha='center');ax.text(4,-.42,'Neighbor $x_j$',ha='center')
    ax.text(.2,2.05,'Nearest majority',va='center',fontsize=10)
    ax.set_title(title,pad=15);ax.set(xlim=(-.45,4.55),ylim=(-.85,2.8));ax.set_aspect('equal');ax.axis('off')
fig.suptitle(r'$z=x_i+\lambda(x_j-x_i)$: same interpolation form, different allowed step',fontsize=18,y=1.02)
fig.text(.5,.05,r'Capped example: $L=4$, $d=2$, $c=0.5$; $\lambda_{max}=\min(1,cd/L)=0.25$.  Rare-branch cap uses $c=0.35$.',ha='center',fontsize=12)
save(fig,'03_interpolation_geometry','Interpolation mechanics',
 'Analytical illustration, not measured data or randomly generated results. The source-neighbor pair is held fixed to isolate the interpolation rule. CASB caps non-safe coverage and border steps at min(1,0.5d/L), and rare steps at min(1,0.35d/L). Safe coverage sources with local majority fraction <=0.2 remain unrestricted. Zero-length edges use lambda=0. CASB-relaxed and CASB-edited use unrestricted interpolation; CASB-ROS sets lambda=0. Blue dots are minority observations, gray is a majority observation, triangles indicate possible synthetic positions.')

# One fixed artificial data example; no tuning to maximize visual separation.
rng=np.random.default_rng(42)
major=rng.normal([0,0],[1.05,.8],size=(180,2))
minor=np.vstack([rng.normal([1.45,.8],[.45,.35],size=(28,2)),rng.normal([-1.7,1.35],[.28,.22],size=(8,2)),[[-.3,-1.5]]])
X=np.vstack([major,minor]);y=np.r_[np.zeros(len(major),int),np.ones(len(minor),int)]
base=CASBSMOTE(ratio=.75,random_state=42);base.fit_resample(X,y)
ids=np.flatnonzero(y==1)
all_nn=NearestNeighbors(n_neighbors=12).fit(X).kneighbors(X[ids],return_distance=False)
neighbors=[row[row!=i][:10] for row,i in zip(all_nn,ids)]
overlap=np.array([np.mean(y[j]==0) for j in neighbors]);support=np.array([np.sum(y[j]==1) for j in neighbors])
same=NearestNeighbors(n_neighbors=7).fit(minor).kneighbors(minor,return_distance=False)
same=[row[row!=i][:5] for i,row in enumerate(same)]
third=np.array([np.linalg.norm(minor[j]-minor[i],axis=1)[2] for i,j in enumerate(same)])
dmajor=NearestNeighbors(n_neighbors=1).fit(major).kneighbors(minor)[0].ravel()
border=(overlap>.2)&(overlap<=.7)&(support>=2)&(dmajor>0)
rare=(third>=np.quantile(third,.75))&(support>=2)&(dmajor>0)
assert border.sum()==base.diagnostics_['border_pool'] and rare.sum()==base.diagnostics_['rare_pool']
fig,axes=plt.subplots(1,3,figsize=(15,5),sharex=True,sharey=True)
for ax,mask,title,color in zip(axes,[np.ones(len(minor),bool),border,rare],['Coverage: every minority source','Border: local majority fraction','Rare: large minority-neighbor distance'],[BLUE,ORANGE,GREEN]):
    ax.scatter(*major.T,c=GRAY,s=16,alpha=.5);ax.scatter(*minor.T,c=BLUE,s=30,alpha=.35)
    ax.scatter(*minor[mask].T,facecolors='none',edgecolors=color,s=120,lw=1.5)
    ax.set_title(title+'\n'+str(int(mask.sum()))+' eligible sources',fontsize=12)
    ax.set_xlabel('Feature 1');ax.set_ylabel('Feature 2');ax.spines[['top','right']].set_visible(False)
fig.suptitle('Overlapping source pools computed by the implemented CASB rules',fontsize=18,y=1.04)
fig.tight_layout()
save(fig,'04_source_pools','Coverage, border and rare sources',
 'Fixed artificial two-dimensional dataset (seed 42), with pool membership computed using the implementation rules and verified against CASB diagnostics. Rings identify eligible sources, not generated observations. Coverage includes all minority sources, even potentially noisy observations. Border requires 0.2<majority fraction<=0.7; rare requires third-minority-neighbor distance >= its 75th percentile. Both require at least two minority neighbors among up to ten full-data neighbors and positive nearest-majority distance. Pool overlap is intentional; a sparse point is not automatically a verified rare subpopulation.')

methods=['original','smote','bsmote','casb','casb_relaxed','casb_edited','casb_ros']
names={'original':'Original','smote':'SMOTE','bsmote':'Borderline-SMOTE','ros':'ROS','adasyn':'ADASYN','casb':'CASB','casb_relaxed':'CASB-relaxed','casb_edited':'CASB-edited','casb_ros':'CASB-ROS'}
fig,axes=plt.subplots(2,4,figsize=(16,8),sharex=True,sharey=True)
toy_audit=[]
for ax,m in zip(axes.flat,methods):
    s=sampler(m,y,42,.75)
    xr,yr=(X.copy(),y.copy()) if s is None else s.fit_resample(X,y)
    rem=np.asarray(getattr(s,'removed_indices_',[]),int);keep=np.ones(len(y),bool);keep[rem[rem<len(y)]]=False
    generated=s.generator_.generated_ if m=='casb_edited' else xr[len(X):]
    ax.scatter(*X[keep&(y==0)].T,c=GRAY,s=16,alpha=.5)
    ax.scatter(*minor.T,c=BLUE,s=28,zorder=3)
    if len(generated):
        if m=='casb_ros':
            unique,count=np.unique(generated,axis=0,return_counts=True)
            ax.scatter(*unique.T,s=45+15*np.sqrt(count),facecolors='none',edgecolors=GREEN,zorder=4)
        else:ax.scatter(*generated.T,c=ORANGE,marker='^',s=21,alpha=.7,zorder=2)
    if len(rem):ax.scatter(*X[rem].T,c=PURPLE,marker='x',s=28)
    ax.set_title(names[m]+f'\nAdded {len(generated)}; removed {len(rem)}',fontsize=12)
    ax.set_xlabel('Feature 1');ax.set_ylabel('Feature 2');ax.spines[['top','right']].set_visible(False)
    toy_audit.append({'method':m,'added':len(generated),'removed':len(rem),'seed':42,'ratio':.75})
axes.flat[-1].axis('off')
axes.flat[-1].legend(handles=[Line2D([],[],marker=mk,color=c,ls='',label=l,markerfacecolor='none' if mk=='o' and c==GREEN else c) for mk,c,l in [('o',GRAY,'Original majority'),('o',BLUE,'Original minority'),('^',ORANGE,'Generated minority'),('o',GREEN,'Added duplicates'),('x',PURPLE,'Removed majority')]],loc='upper left',frameon=False)
fig.suptitle('Resampling comparison',fontsize=20,y=1.01);fig.tight_layout()
save(fig,'05_controlled_resampling','Controlled resampling example',
 'Actual executions of the shared samplers on the artificial dataset of Figure 4. All methods use target ratio 0.75 and seed 42, isolating generation and cleaning behavior from ratio tuning. New samples are identified through sampler provenance; CASB-edited uses generator_.generated_ because ENN reorders output rows. CASB-ROS ring sizes reflect duplicate multiplicity. This is an illustrative case, not evidence of superior classification or noise handling. No UGO results are fabricated. All original minority samples are retained by CASB-edited.')

# Verify and aggregate existing benchmark evidence.
rows=[];inputs={}
for p in sorted(ROOT.glob('[0-9][0-9]_*_CASB.ipynb')):
    nb=json.loads(p.read_text(encoding='utf8'));meta=nb['metadata']['current_comparison']
    assert meta['complete'] and len(meta['records'])==360
    for src,digest in meta['input_sha256'].items():
        path=ROOT/src
        if not path.exists():path=ROOT/'output'/src
        assert hashlib.sha256(path.read_bytes()).hexdigest()==digest
    inputs[p.name]=hashlib.sha256(p.read_bytes()).hexdigest()
    rows.extend(meta['records'])
frame=pd.DataFrame(rows);bad=sorted(set(frame.loc[frame.error.notna(),'dataset']))
valid=frame[~frame.dataset.isin(bad)]
assert len(frame)==15840 and len(valid)==14400 and valid.dataset.nunique()==40
metrics=['gmean','f1','precision','recall'];family=['casb','casb_relaxed','casb_edited','casb_ros'];core=['smote','bsmote','ros','adasyn']
means=valid.groupby(['dataset','classifier','method'])[metrics].mean()
ds=means.groupby(['dataset','method']).mean()
CLASSIFIERS=['svm','knn','rf','lr']
views={c.upper():means.xs(c,level='classifier') for c in CLASSIFIERS}
views['SVM + KNN + RF + LR']=ds
avg=ds.groupby('method').mean()
fig,axes=plt.subplots(2,2,figsize=(13,9))
for ax,metric in zip(axes.flat,metrics):
    delta=100*(avg.loc[family,metric].to_numpy()[:,None]-avg.loc[core,metric].to_numpy()[None,:])
    vmax=5
    im=ax.imshow(delta,cmap='RdBu',vmin=-vmax,vmax=vmax)
    ax.set_xticks(range(4),['SMOTE','B-SMOTE','ROS','ADASYN'],fontsize=10)
    ax.set_yticks(range(4),[names[m] for m in family],fontsize=10);ax.set_title(metric.upper() if metric=='f1' else metric.capitalize())
    for (i,j),v in np.ndenumerate(delta):ax.text(j,i,f'{v:+.2f}',ha='center',va='center',color='white' if abs(v)>3 else NAVY,fontsize=12)
fig.suptitle('Mean score differences: CASB variant minus baseline',fontsize=20)
fig.subplots_adjust(top=.89,bottom=.12,hspace=.42,wspace=.35,right=.88)
cax=fig.add_axes([.91,.2,.018,.6]);fig.colorbar(im,cax=cax,label='Percentage points; positive favors CASB')
fig.text(.5,.035,'40 complete datasets; equal dataset and classifier weights. Descriptive averages, not significance tests.',ha='center',fontsize=11)
save(fig,'06_average_score_differences','Average benchmark differences',
 'B-SMOTE denotes Borderline-SMOTE. Differences in equal-dataset-weight mean scores on 40 complete datasets. Each dataset/method/classifier first averages five supplied folds and two seeds; SVM, KNN, RF and LR then receive equal weight. Four datasets with 52 ADASYN failures are excluded from all methods in this matched comparison. No best-of-family maximum or composite metric is tested or displayed. Positive cells favor the named fixed CASB variant. These averages do not establish statistical significance.')

fig,axes=plt.subplots(2,2,figsize=(15,10))
order=core+family
for ax,metric in zip(axes.flat,metrics):
    for offset,(view,color) in zip([-.30,-.10,.10,.30],[('SVM',BLUE),('KNN',ORANGE),('RF',GREEN),('LR',PURPLE)]):
        wide=views[view][metric].unstack('method')[order]
        ranks=wide.rank(axis=1,ascending=False,method='average')
        bp=ax.boxplot([ranks[m] for m in order],positions=np.arange(8)+offset,widths=.16,patch_artist=True,showfliers=False,manage_ticks=False)
        for p in bp['boxes']:p.set_facecolor(color);p.set_alpha(.5)
        for med in bp['medians']:med.set_color(NAVY)
    ax.set_xticks(range(8),[names[m] for m in order],rotation=35,ha='right',fontsize=10)
    ax.set_ylim(.5,8.5);ax.set_yticks(range(1,9));ax.set_ylabel('Dataset rank (1 = best)');ax.set_title(metric.capitalize());ax.grid(axis='y',alpha=.15)
fig.suptitle('Rank distributions across the 40 matched datasets',fontsize=21)
fig.legend(handles=[Line2D([],[],color=BLUE,lw=8,label='SVM'),Line2D([],[],color=ORANGE,lw=8,label='KNN'),Line2D([],[],color=GREEN,lw=8,label='RF'),Line2D([],[],color=PURPLE,lw=8,label='LR')],loc='upper center',bbox_to_anchor=(.5,.95),ncol=4,frameon=False)
fig.tight_layout(rect=(0,.02,1,.90))
save(fig,'07_rank_distributions','Dataset-level rank distributions',
 'Ranks of eight resampling methods, computed separately for each classifier and metric from the per-dataset fold/seed means. Lower ranks are better, tied scores receive average ranks. Boxes represent the interquartile range, lines the median, whiskers extend to 1.5 IQR; outlier markers are omitted for readability. Ranks concern 40 datasets, not 400 independent folds. Related source-dataset variants are correlated. The no-resampling reference is not included in these eight-method ranks.')

# Classifier-specific independently reproduced tests.
verified=pd.read_csv(ROOT/'statistical result/wilcoxon_holm_results.csv')
verified=verified[verified.classifier.isin(CLASSIFIERS)]
paired=verified[verified.analysis=='dataset_pairs'].copy()
grouped=verified[verified.analysis=='source_family_mean'].copy()
assert len(paired)==384
fig,axes=plt.subplots(2,2,figsize=(16,11))
for ax,clf in zip(axes.flat,CLASSIFIERS):
    q=paired[paired.classifier==clf];baselines=list(dict.fromkeys(q.baseline))
    matrix=np.zeros((16,len(baselines)))
    for i,(metric,method) in enumerate((metric,method) for metric in metrics for method in family):
        for j,b in enumerate(baselines):
            r=q[(q.metric==metric)&(q.method==method)&(q.baseline==b)].iloc[0]
            matrix[i,j]=r.rank_biserial if r.p_holm<.05 else 0
            ax.text(j,i,'+' if matrix[i,j]>0 else '-' if matrix[i,j]<0 else '.',ha='center',va='center',fontsize=9)
    ax.imshow(matrix,cmap='RdBu',vmin=-1,vmax=1,aspect='auto')
    ax.set_xticks(range(len(baselines)),[names.get(b,b.upper()) for b in baselines],rotation=35,ha='right',fontsize=9)
    ax.set_yticks(range(16),[metric+' / '+names[method] for metric in metrics for method in family],fontsize=8)
    ax.set_title(clf.upper()+': '+str(int(q.correction_n.iloc[0]))+' corrected comparisons')
fig.suptitle('Paired fixed-variant significance by classifier',fontsize=20)
fig.text(.5,.015,'+ significant benefit; - significant harm; . not significant. Color: paired rank-biserial effect.',ha='center',fontsize=11)
fig.tight_layout(rect=(0,.045,1,.95))
save(fig,'08_adjusted_significance','Classifier-specific adjusted paired comparisons',
 'Two-sided Wilcoxon tests reproduced independently in statistical result. Holm correction covers 112 fixed-variant comparisons across four metrics for SVM/KNN, including no resampling, RUS and SMOTE-ENN; RF/LR use 80 comparisons including no resampling. Classifiers are kept separate. Pairwise units are 44 complete datasets, or 40 for ADASYN. Plus: significant CASB benefit; minus: significant harm; dot: non-significant at 0.05. Color encodes rank-biserial effect. Source-family sensitivity and exact sign tests are in the statistical report. No tests apply to the retrospective family maximum. Classifier retention was post hoc; findings are exploratory.')

display(pd.DataFrame(toy_audit));display(avg.loc[core+family]*100)
with pd.option_context('display.max_rows',200,'display.max_columns',20):
    display(paired);display(grouped)
display(paired[paired.p_holm<.05]);display(grouped[grouped.p_holm<.05])
print('Pairwise significant comparisons:',int((paired.p_holm<.05).sum()))
print('Source-family paired units:',grouped.n.min(),grouped.n.max(),'; significant:',int((grouped.p_holm<.05).sum()))
print('Source-family exact sign-test significant:',int((grouped.sign_p_holm<.05).sum()))
print('Excluded matched-cohort datasets:',bad)
display(pd.DataFrame({'notebook':list(inputs),'sha256':list(inputs.values())}))
print('Shared implementation SHA256:',hashlib.sha256((ROOT/'binary_oversampling.py').read_bytes()).hexdigest())

# Figure captions, evidence limits and editable manuscript text.
intro='''# CASB paper figures and captions

All diagrams are original vector drawings; empirical figures use the saved project results. No UGO artwork or UGO experiment outputs have been copied. PNG (300 DPI), PDF and editable SVG versions are provided. SVG text requires DejaVu Sans or a compatible font. The executed notebook contains provenance checks and statistical tables. No watermark or branding is added.

## Relationship to the reference paper

The uploaded UGO paper uses noise illustrations (Fig. 1), sampling/workflow diagrams (Figs. 2-4), an iteration example and resampled distributions (Figs. 5-6), simulated-noise studies (Figs. 7 and 9-12), dataset descriptions (Fig. 8), rank plots (Fig. 13), and pairwise statistical comparisons (Fig. 14). Our Figures 1-5 serve the mechanism/explanation roles, and Figures 6-8 report our existing benchmark evidence. UGO uses classifier uncertainty, repeated retraining, noise pre-detection and synthetic-point relocation. CASB does not implement those operations. The separate controlled-overlap notebook supplies Figures 15-20; it varies geometric overlap, not corrupted-label percentage. No uncertainty-reduction experiment is claimed.

The existing real-dataset PCA figures remain in `paper plot/`. They are selected positive examples, not representative samples of all datasets. The new controlled Figure 5 uses a common ratio to separate generation differences from hyperparameter tuning.

## Classifier selection

RF and LR were retained after examining benchmark results. GaussianNB results, including its losses, remain in progress.md. The four-classifier presentation and significance tests are exploratory and do not adjust away this post-hoc classifier selection.

## Manuscript interpretation

CASB is a SMOTE-based family combining overlapping source pools with generation choices. CASB-edited adds established majority-only ENN cleaning. Novelty of the combination remains to be established against close prior work; lower apparent overlap in a plot is not a classification-performance or noise-removal proof. CASB's coverage pool can still draw noisy minority sources. Statistical analysis is retrospective and the dataset variants are not independent replications. The source-family sensitivity analysis is included for that reason; non-significance is not evidence of equivalence.

'''
intro += f'## Statistical findings\n\nAcross {len(paired)} classifier-specific comparisons, {int((paired.p_holm<.05).sum())} show significant differences in either direction. Paired comparisons use 44 valid datasets, or 40 for ADASYN. Mean/rank figures use the strict common 40-dataset cohort. Holm families contain 112 tests for SVM/KNN and 80 for RF/LR. Source-family sensitivity uses 12-14 units and has {int((grouped.p_holm<.05).sum())} significant comparisons. Corrected source-family sign tests have {int((grouped.sign_p_holm<.05).sum())} rejections. These results do not support blanket superiority.\n\n'
text=intro+'\n'.join(f'## Figure {i}. {title}\n\n{caption}\n' for i,(title,caption) in enumerate(captions,1))
(OUT/'Figure_Captions.md').write_text(text,encoding='utf8')

from reportlab.pdfgen import canvas
from reportlab.lib.utils import ImageReader
from reportlab.platypus import Paragraph
from reportlab.lib.styles import ParagraphStyle
from reportlab.lib import colors
from xml.sax.saxutils import escape
from PIL import Image as PILImage
book=OUT/'CASB_Paper_Figure_Atlas.pdf';c=canvas.Canvas(str(book),pagesize=(1000,780))
c.setTitle('CASB: Method and Benchmark Figures');c.setAuthor('')
style=ParagraphStyle('caption',fontName='Helvetica',fontSize=11,leading=15,textColor=colors.HexColor(NAVY))
for i,(stem,(title,caption)) in enumerate(zip(figures,captions),1):
    c.setFont('Helvetica-Bold',17);c.setFillColor(colors.HexColor(NAVY));c.drawString(38,747,f'Figure {i}. {title}')
    im=PILImage.open(OUT/(stem+'.png'));w,h=im.size;scale=min(924/w,540/h);dw,dh=w*scale,h*scale
    c.drawImage(ImageReader(im),(1000-dw)/2,190+(540-dh)/2,width=dw,height=dh)
    p=Paragraph(escape(caption),style);_,ph=p.wrap(924,150);assert ph<154,(i,ph)
    p.drawOn(c,38,170-ph)
    c.setFont('Helvetica',9);c.drawRightString(962,20,str(i));c.showPage()
c.save()
print('Saved eight figures in PNG/PDF/SVG, caption notes and figure atlas.')
```
